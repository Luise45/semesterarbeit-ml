```python
import ee
import geemap
```

```python
ee.Authenticate()
```

```python
ee.Initialize(project='ds-projekt-509618')
```

```python
lat = 48.20
lon = 11.50
ee.Geometry.Point([lat, lon])

point = ee.Geometry.Point([lon, lat])
```

```python
date = ee.Date('2024-06-15')

collection = (
    ee.ImageCollection('MODIS/061/MOD13Q1')
    .filterDate(date, date.advance(16, 'day'))
    .filterBounds(point)
)
image = ee.Image(collection.first())

ndvi_raw = image.select('NDVI').reduceRegion(
    reducer=ee.Reducer.first(),
    geometry=point,
    scale=250
).get('NDVI')

print(ndvi_raw.getInfo())
ndvi = ee.Number(ndvi_raw).multiply(0.0001)

print("NDVI:", ndvi.getInfo())
```

```python
image = ee.Image(collection.first())

actual_date = ee.Date(
    image.get('system:time_start')
).format('YYYY-MM-dd')

ndvi_raw = image.select('NDVI').reduceRegion(
    reducer=ee.Reducer.first(),
    geometry=point,
    scale=250
).get('NDVI')

ndvi = ee.Number(ndvi_raw).multiply(0.0001)

print("Koordinate:", lat, lon)
print("Gewünschtes Datum: 2024-06-15")
print("MODIS-Datum:", actual_date.getInfo())
print("NDVI:", ndvi.getInfo())
```

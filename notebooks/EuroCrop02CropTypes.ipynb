```python
# fint the crop codes 
import polars as pl

codes = (
    fields
    .group_by("original_code")
    .agg(
        pl.len().alias("number_of_fields"),
        pl.col("area_ha").mean().alias("mean_area_ha")
    )
    .sort("number_of_fields", descending=True)
    .collect()
)

print(codes)
```
